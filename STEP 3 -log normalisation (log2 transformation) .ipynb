```python
import pandas as pd

# 1. Load your log-normalized file
df= pd.read_excel(' FILE 1 teak working genes con_sorted_(2).xlsx')

# 2. Identify the numeric columns (skipping the Gene_ID column)
numeric_cols = df.columns[1:]

# 3. Filter out rows where the sum of all sample values is 0
# This keeps only genes that are expressed in at least one sample
df_cleaned = df[df[numeric_cols].sum(axis=1) > 0]




print(f"Original genes: {len(df)}")
print(f"Genes removed: {len(df) - len(df_cleaned)}")
print(f"Remaining genes: {len(df_cleaned)}")
```

```python
df_cleaned.to_excel("ExpressionNew.xlsx",index=False)
```

```python
num_cols=df_cleaned.select_dtypes(include="number").columns
print("numeric columns")
print(num_cols)
df_cleaned[num_cols]=np.log2(df_cleaned[num_cols]+1)
print(df_cleaned.head())
```

```python
df.to_excel('EXPRESSEDGENES0 GENES REMOVES AND LOG NORMALISED.xlsx'),index=False)
```
